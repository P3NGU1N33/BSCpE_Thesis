Create A Training Climatology Table
- Using the existing iSWAI_HistoryData.csv to compute typical values by month and hour
- This gives you a “normal expected temp/humidity/etc” for each hour of the year.

```python
import pandas as pd

HIST_PATH = "../dataset/iSWAI_HistoryData.csv"  # adjust
OUT_PATH = "../server/data/climatology_month_hour.csv"

df = pd.read_csv(HIST_PATH)
df["datetime"] = pd.to_datetime(df["datetime"], errors="coerce")
df = df.dropna(subset=["datetime"])

df["month"] = df["datetime"].dt.month
df["hour"] = df["datetime"].dt.hour
df 
```

```python
weather_cols = ["temp","humidity","precip","windgust","windspeed", "winddir","sealevelpressure","cloudcover","visibility"]
weather_cols = [c for c in weather_cols if c in df.columns]
weather_cols  
```

Grouping all the months and hours by getting each mean of column.

```python
clim = (
    df.groupby(["month","hour"], as_index=False)[weather_cols]
      .mean()
)
```

```python
clim
```

```python
clim.to_csv(OUT_PATH, index=False)
print("Saved:", OUT_PATH, "rows:", len(clim))
```

Add DateTime Cyclical Features

```python
import numpy as np  
def add_datetime_features(df):
    df = df.copy()
    df["datetime"] = pd.to_datetime(df["datetime"], errors="coerce")
    df = df.sort_values("datetime").reset_index(drop=True)

    df["hour"] = df["datetime"].dt.hour
    df["dayofweek"] = df["datetime"].dt.dayofweek
    df["month"] = df["datetime"].dt.month
    df["day"] = df["datetime"].dt.day

    df["hour_sin"] = np.sin(2*np.pi*df["hour"]/24)
    df["hour_cos"] = np.cos(2*np.pi*df["hour"]/24)
    df["month_sin"] = np.sin(2*np.pi*df["month"]/12)
    df["month_cos"] = np.cos(2*np.pi*df["month"]/12)
    return df

df_2026 = add_datetime_features(df_2026)
df_2026 = df_2026.sort_values("datetime").reset_index(drop=True)
df_2026
```

Recursive wind speed direction

```python
import joblib
import numpy as np

ws = joblib.load("../server/models/windspeed_model.joblib")
rf = ws["rf"]
xgb_model = ws["xgb"]
feature_cols = ws["feature_cols"]
w = ws.get("weights", {"rf": 0.5, "xgb": 0.5})

def add_lag_roll_features(df) :
    df = df.copy()
    df = df.sort_values("datetime").reset_index(drop=True)

    # Lags for windspeed and windgust
    wind_lags = [1, 2, 3, 6, 12, 24]
    gust_lags = [1, 2, 3, 6, 12, 24]

    for lag in wind_lags:
        df[f"wind_lag_{lag}"] = df["windspeed"].shift(lag)

    for lag in gust_lags:
        df[f"gust_lag_{lag}"] = df["windgust"].shift(lag)

    # Rolling stats (use previous values, shift(1) to prevent leakage)
    windows = [3, 6, 12, 24]
    for w in windows:
        df[f"wind_roll_mean_{w}"] = df["windspeed"].shift(1).rolling(w).mean()
        df[f"wind_roll_std_{w}"]  = df["windspeed"].shift(1).rolling(w).std()

    return df

# start with existing windspeed (ESP32) if available, else NaN
df_2026["windspeed"] = pd.to_numeric(df_2026.get("windspeed"), errors="coerce")
df_2026 = add_lag_roll_features(df_2026)

for i in range(len(df_2026)):
    if pd.notna(df_2026.loc[i, "windspeed"]):
        continue

    row = df_2026.loc[i, feature_cols]
    if row.isna().any():
        continue

    X = row.to_numpy().reshape(1, -1)
    pred = w["rf"] * rf.predict(X)[0] + w["xgb"] * xgb_model.predict(X)[0]
    df_2026.loc[i, "windspeed"] = float(pred)

    # update lags
    df_2026 = add_lag_roll_features(df_2026)    
```
